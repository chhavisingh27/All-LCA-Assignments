```python

def read_matrix(rows, cols, name):
    print(f"\nEnter elements of matrix {name}:")
    matrix = []
    for i in range(rows):
        row = []
        for j in range(cols):
            row.append(int(input(f"{name}[{i}][{j}]: ")))
        matrix.append(row)
    return matrix


def add_matrices(A, B, rows, cols):
    result = []
    for i in range(rows):
        row = []
        for j in range(cols):
            row.append(A[i][j] + B[i][j])
        result.append(row)
    return result


def display_matrix(matrix, title):
    print(f"\n{title}")
    for row in matrix:
        print(" ".join(f"{val:5}" for val in row))


rows = int(input("Enter number of rows: "))
cols = int(input("Enter number of columns: "))

A = read_matrix(rows, cols, "A")
B = read_matrix(rows, cols, "B")

C = add_matrices(A, B, rows, cols)

display_matrix(A, "Matrix A:")
display_matrix(B, "Matrix B:")
display_matrix(C, "Sum of A and B:")
```
